```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
pollut1=pd.read_csv('pollut_20261012084158 - 2025-01-01 00_00 - 2025-12-31 23_00.csv', sep=';')
pollut2=pd.read_csv('pollut_20261012084855 - 2025-01-01 00_00 - 2025-12-31 23_00.csv', sep=';')
pollut3=pd.read_csv('pollut_20261012084932 - 2025-01-01 00_00 - 2025-12-31 23_00.csv', sep=';')
pollut4=pd.read_csv('pollut_20261012085004 - 2025-01-01 00_00 - 2025-12-31 23_00.csv', sep=';')
pollut5=pd.read_csv('pollut_20261012085038 - 2025-01-01 00_00 - 2025-12-31 23_00.csv', sep=';')
```

```python
pollut=pd.concat([pollut1,pollut2, pollut3, pollut4, pollut5], axis=0)

display(pollut.head(5))
display(pollut.tail(5))

print(pollut.shape)
```

```python
# Recherche d'une colonne qui contiendrait l'information du département
dep=pollut['code site'].unique()
dep=sorted(dep)
print(dep)
# ECHEC => demande à Claude qui propose la solution d'interroger l'API gouvernementale qui associe longitude/latitude à un département
# ECHEC supplémentaire quand je réalise qu'il manque des mesures pour des départements entiers => retour à la régionalisation
```

```python
import requests

insee_to_iso = {
    '84': 'FR-ARA', '27': 'FR-BFC', '53': 'FR-BRE', '24': 'FR-CVL',
    '94': 'FR-20R', '44': 'FR-GES', '32': 'FR-HDF', '11': 'FR-IDF',
    '28': 'FR-NOR', '75': 'FR-NAQ', '76': 'FR-OCC', '52': 'FR-PDL',
    '93': 'FR-PAC',
    '01': 'FR-GUA', '02': 'FR-MQ',  '03': 'FR-GF',
    '04': 'FR-RE',  '06': 'FR-MYT',
}

def get_region(lat, lon):
    url = f"https://geo.api.gouv.fr/communes?lat={lat}&lon={lon}&fields=region"
    r = requests.get(url)
    if r.ok and r.json():
        code_insee = r.json()[0]['region']['code']
        return insee_to_iso.get(code_insee, None)
    return None

sites = pollut[['code site', 'Latitude', 'Longitude']].drop_duplicates('code site')
sites['region'] = sites.apply(
    lambda r: get_region(r['Latitude'], r['Longitude']), axis=1
)

pollut = pollut.merge(sites[['code site', 'region']], on='code site', how='left')
```

```python
display(pollut.head(5))
display(pollut.tail(5))

print(pollut.shape)
# Le dataset contient maintenant l'information nécessaire en chiffre et en lettre, placé tout à la fin du dataframe
```

```python
pollut = pollut.reset_index(drop=True)
pollut['valeur'] = pollut['valeur'].where(pollut['valeur'] >= 0)

print(pollut.columns.tolist())
print(pollut["Polluant"].value_counts())

# Les polluants SO2 et NO2 ne sont pas assez représentés => à éliminer du dataset final
pollut=pollut[pollut["Polluant"]!='NO2']
pollut=pollut[pollut["Polluant"]!='SO2']

polluants=pollut["Polluant"].unique().tolist()
```

```python
print(pollut["type d'influence"].value_counts())
pollut_avant = pollut.copy()
pollut = pollut[pollut["type d'influence"] == 'Fond']
#“Fond” donne le niveau de pollution de fond régional uniquement

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
categories = list(set(pollut_avant["type d'implantation"]).union(set(pollut["type d'implantation"])))
palette = plt.cm.tab10.colors
couleurs = {cat: palette[i] for i, cat in enumerate(categories)}

# Avant
counts_avant = pollut_avant["type d'implantation"].value_counts()
axes[0].pie(counts_avant.values,
            labels=counts_avant.index,
            colors=[couleurs[c] for c in counts_avant.index],
            autopct='%1.1f%%')
axes[0].set_title("Tous types d'influence")

#type d'implantation — distingue Urbaine, Rurale nationale, Rurale près des villes, etc. 
#Mieux équilibré après sélection de "Fond" régional uniquement

counts_apres = pollut["type d'implantation"].value_counts()
axes[1].pie(counts_apres.values,
            labels=counts_apres.index,
            colors=[couleurs[c] for c in counts_apres.index],
            autopct='%1.1f%%')
axes[1].set_title("Types d'influence = Fond uniquement)")

plt.show()
```

```python
pollut['Date de fin'] = pd.to_datetime(pollut['Date de fin'])

pollut['week'] = None
mask_2025 = pollut['Date de fin'].dt.year == 2025
pollut.loc[mask_2025, 'week'] = pollut.loc[mask_2025, 'Date de fin'].dt.isocalendar().week

temp = pollut.pivot_table(
    index=['region', 'week'],
    columns='Polluant',
    values='valeur'
).reset_index()

display(temp.head(5))

regions = ['FR-IDF','FR-NOR', 'FR-PAC']
pol = 'C6H6'

fig, ax = plt.subplots(figsize=(12, 4))

for region in regions:
    serie = temp[temp['region'] == region].sort_values('week')
    serie = serie[['week', pol]].dropna()
    ax.plot(serie['week'], serie[pol], marker='o', markersize=3, label=region)

ax.set_xlabel('Semaine')
ax.set_ylabel(pol)
ax.legend()
ax.grid(True)
plt.tight_layout()
plt.show()
```

```python
temp.groupby('region')[polluants].count()
```

```python
pollut_region=temp.copy()
pollut_region_interp = pollut_region.copy()

for region in pollut_region['region'].unique():
    mask = pollut_region_interp['region'] == region
    pollut_region_interp.loc[mask, polluants] = (
        pollut_region_interp.loc[mask, polluants]
        .interpolate(method='linear', limit_area='inside')
    )
    
print(pollut_region_interp.isna().sum())
# Malgré la présence de lignes entièrement NA, conservons tout le tableau de référence

#pollut_region_sansna = pollut_region_interp.dropna(subset=polluants, how='all')
#print(pollut_region_sansna.isna().sum())
#display(pollut_region_sansna.head(5))
```

```python
region_choisie = 'FR-PAC'

corr = (
    pollut_region_interp[pollut_region_interp['region'] == region_choisie][polluants]
    .corr()
)

fig, ax = plt.subplots(figsize=(6, 6))

im = ax.imshow(corr.values, cmap='RdBu_r', aspect='auto', vmin=-1, vmax=1)
plt.colorbar(im, ax=ax)

ax.set_xticks(range(len(polluants)))
ax.set_xticklabels(polluants, rotation=45, ha='right')
ax.set_yticks(range(len(polluants)))
ax.set_yticklabels(polluants)

for i in range(len(polluants)):
    for j in range(len(polluants)):
        ax.text(j, i, f'{corr.values[i, j]:.2f}', ha='center', va='center', fontsize=8)

ax.set_title(f'Corrélations entre polluants — {region_choisie}')
plt.tight_layout()
plt.show()
```
