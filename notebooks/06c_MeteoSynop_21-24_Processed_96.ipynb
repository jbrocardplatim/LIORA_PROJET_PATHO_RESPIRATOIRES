Quelques modifications sur le fichier final pour harmoniser avec les autres fichiers avant le merge 

```python
import pandas as pd
import numpy as np
```

```python
df_meteo_hebdo_clean_96 = pd.read_csv("../data/processed/df_meteo_hebdo_clean_96.csv")

df_meteo_hebdo_clean_96.info()
```

```python
df_meteo_hebdo_clean_96.head(10)
```

```python
#remplace le w en s pour les dates de la colonne annee_semaine
df_meteo_hebdo_clean_96['annee_semaine'] = df_meteo_hebdo_clean_96['annee_semaine'].str.replace('W', 'S', regex=False)

df_meteo_hebdo_clean_96.info()

df_meteo_hebdo_clean_96.head(10)
```

```python
#retire les colonnes latitude et longitude car plus utilisées
df_meteo_hebdo_clean_96 = df_meteo_hebdo_clean_96.drop(columns=['latitude', 'longitude'], errors='ignore')

df_meteo_hebdo_clean_96.info()

df_meteo_hebdo_clean_96.head(10)
```

```python
#rename the 'annee_semaine' column to 'date_semaine'
df_meteo_hebdo_clean_96 = df_meteo_hebdo_clean_96.rename(columns={'annee_semaine': 'date_semaine'})

df_meteo_hebdo_clean_96.info()

df_meteo_hebdo_clean_96.head(10)
```

```python
# Conserver uniquement les lignes qui ne contiennent pas '2020-S53' ni '2025-S01'
# Identifier les index des lignes à supprimer
index_to_drop = df_meteo_hebdo_clean_96[df_meteo_hebdo_clean_96['date_semaine'].isin(['2020-S53', '2025-S01'])].index

# Supprimer ces lignes du DataFrame
df_meteo_hebdo_clean_96 = df_meteo_hebdo_clean_96.drop(index=index_to_drop)

df_meteo_hebdo_clean_96.info()

df_meteo_hebdo_clean_96.head(10)

display(df_meteo_hebdo_clean_96['date_semaine'].unique())

```

```python
#exporte le dataframe final_meteo_hebdo_clean_96 en csv
df_meteo_hebdo_clean_96.to_csv("../data/final_metropole/final_meteo_hebdo_clean_96.csv", index=False, encoding='utf-8')
```
