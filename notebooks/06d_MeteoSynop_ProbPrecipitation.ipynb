```python
import pandas as pd
import plotly.express as px

# Supposons que ton DataFrame source s'appelle df et contienne tes colonnes de précipitations
# Si tes données sont au format large (une colonne par variable), on les passe au format long :
cols_precip = [
    'precipitation_1h', 
    'precipitation_3h', 
    'precipitation_6h', 
    'precipitation_12h', 
    'precipitation_24h'
]

df_long = Merge_final.melt(
    value_vars=cols_precip, 
    var_name='Variables de précipitation', 
    value_name='Cumul hebdomadaire (mm)'
)

# Création du boxplot interactif avec Plotly Express
fig = px.box(
    df_long,
    x='Cumul hebdomadaire (mm)',
    y='Variables de précipitation',
    orientation='h',
    color='Variables de précipitation',
    title='Distribution des cumuls hebdomadaires de précipitations',
    template='plotly_white'
)

# Personnalisation de l'affichage
fig.update_layout(
    xaxis_title='Cumul hebdomadaire (mm)',
    yaxis_title='Variables de précipitation',
    showlegend=False
)

# Affichage du graphique interactif (dans un Jupyter Notebook ou export HTML)
fig.show()

# Optionnel : pour l'exporter en fichier HTML interactif
# fig.write_html("boxplot_interactif.html")
```
